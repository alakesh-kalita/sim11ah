```python
import tensorflow as tf

# Force execution specifically on your M4 GPU
with tf.device('/GPU:0'):
    a = tf.random.normal([2000, 2000])
    b = tf.random.normal([2000, 2000])
    
    # Perform a massive matrix multiplication
    c = tf.matmul(a, b)

print("Calculation successful! M4 Metal GPU is working perfectly.")
```

```python
import tensorflow as tf

# Check if TensorFlow detects the Metal Performance Shaders (MPS) as a GPU
gpu_devices = tf.config.list_physical_devices('GPU')

if gpu_devices:
    DEVICE = "GPU (Apple Metal / MPS)"
else:
    DEVICE = "CPU"

print(f" ⚡ Running on Device: {DEVICE}")
```
